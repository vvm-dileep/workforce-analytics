```python
con.execute("""
SELECT employee_name, department, salary,
       AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
FROM '../data/employees.csv'
LIMIT 10
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, department_avg_salary,
FROM employee_salary
WHERE salary > department_avg_salary
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, department_avg_salary, salary - department_avg_salary AS salary_variance
FROM employee_salary
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, department_avg_salary, 
salary - department_avg_salary AS salary_variance,
((salary - department_avg_salary) / department_avg_salary )* 100 AS salary_variance_pct
FROM employee_salary
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, department_avg_salary, 
salary - department_avg_salary AS salary_variance,
ROUND(
    (salary - department_avg_salary) / department_avg_salary * 100,
    1
) AS salary_variance_pct
FROM employee_salary
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
    SELECT employee_name, department, salary,
           RANK() OVER (PARTITION BY department ORDER BY salary DESC) as salary_rank
    FROM '../data/employees.csv'
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           RANK() OVER (PARTITION BY department ORDER BY salary DESC) as salary_rank
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, salary_rank
FROM employee_salary
Where salary_rank = 1
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) as salary_rank
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, salary_rank
FROM employee_salary
Where salary_rank = 1
ORDER BY salary DESC
""").df()
```

```python
con.execute("""
WITH employee_salary AS (
    SELECT employee_name, department, salary,
           ROW_NUMBER() OVER (
               PARTITION BY department
               ORDER BY salary DESC
           ) AS salary_rank
    FROM '../data/employees.csv'
)
SELECT employee_name, department, salary, salary_rank
FROM employee_salary
ORDER BY department, salary_rank
""").df()
```
